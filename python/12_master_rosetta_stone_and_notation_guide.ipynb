## 2. Notation Collision Survival Guide

When reading interdisciplinary literature, statistical libraries (PyMC, ArviZ,
SciPy), or textbooks (Cover & Thomas vs McElreath vs MacKay), several
mathematical symbols collide disastrously. Use this guide to de-conflict them:

### Collision 1: The Letter $H$
- $H(X)$: **Shannon Entropy** (Information Theory) -- scalar in bits or nats.
- $\mathbf{H}$ or $H_{ij}$: **Hessian Matrix** (Calculus / Bayesian Laplace) --
  matrix of second-order partial derivatives $\frac{\partial^2 \ln p}{\partial \theta_i \partial \theta_j}$.
- $H_0, H_1$: **Null and Alternative Hypotheses** (Decision Theory / Wald SPRT).
- $h(X)$: **Differential Entropy** of a continuous random variable.

### Collision 2: The Letter $I$
- $I(x)$: **Self-Information / Surprise** of an outcome $x$ (scalar bits).
- $I(X; Y)$: **Mutual Information** between random variables $X$ and $Y$ (scalar bits).
- $\mathcal{I}(\boldsymbol{\theta})$: **Fisher Information Matrix** (Information
  Geometry) -- positive semi-definite Riemannian metric tensor.
- $\mathbf{I}$: **Identity Matrix** ($\mathbf{I}_n$).
- $I_x(a, b)$: **Regularized Incomplete Beta Function** (Cumulative Beta CDF).

### Collision 3: The Letter $D$
- $D_{\text{KL}}(P \parallel Q)$: **Kullback-Leibler Divergence** (Relative Entropy).
- $\mathcal{D}$: **Observed Dataset** ($\{\mathbf{y}_1, \dots, \mathbf{y}_n\}$).
- $R(D)$: **Rate-Distortion Function** ($D$ is the maximum distortion budget).
- $d$: **Experimental Design Action** (Lindley EIG).

### Collision 4: The Greek Letter $\beta$
- $\text{Beta}(\alpha, \beta)$: Second shape parameter of the Beta distribution
  (prior pseudo-successes / failures).
- $\beta$ in Information Bottleneck: The Lagrange multiplier $\min I(X; T) - \beta I(T; Y)$
  controlling the compression-retention exchange rate.
- $\beta$ in MCMC / Physics: Inverse temperature $\beta = \frac{1}{k_B T}$ used
  in simulated annealing and parallel tempering.
- $\beta_{\text{err}}$: Type II error probability (probability of merging a real
  regression) in Wald SPRT.

### Collision 5: The Greek Letter $\theta$
- $\theta$: Bernoulli success / failure probability ($\theta \in [0, 1]$).
- $\boldsymbol{\theta}$: General parameter vector ($\boldsymbol{\theta} \in \mathbb{R}^D$).
- $\Theta_0, \Theta_1$: Discrete hypothesis parameter regions in decision theory.


## 3. Master Formula Cheat Sheet

### Part A: Information-Theoretic Foundations

$$
\begin{aligned}
\text{Self-Information (Surprise):} \quad & I(x) = -\log_2 P(x) \quad \text{[bits]} \\
\text{Shannon Entropy:} \quad & H(X) = -\sum_{x} p(x) \log_2 p(x) \\
\text{Binary Entropy:} \quad & H_2(\theta) = -\theta \log_2(\theta) - (1-\theta) \log_2(1-\theta) \\
\text{Relative Entropy (KL):} \quad & D_{\text{KL}}(P \parallel Q) = \sum_{x} P(x) \log_2 \frac{P(x)}{Q(x)} \ge 0 \\
\text{Cross-Entropy:} \quad & H(P, Q) = H(P) + D_{\text{KL}}(P \parallel Q) \\
\text{Mutual Information:} \quad & I(X; Y) = H(X) - H(X \mid Y) = H(Y) - H(Y \mid X) \\
\text{Fisher Information Matrix:} \quad & \mathcal{I}_{ij}(\boldsymbol{\theta}) = \mathbb{E}\left[ \left(\frac{\partial \ln p}{\partial \theta_i}\right) \left(\frac{\partial \ln p}{\partial \theta_j}\right) \right] = - \mathbb{E}\left[ \frac{\partial^2 \ln p}{\partial \theta_i \partial \theta_j} \right] \\
\text{Cramér-Rao Lower Bound:} \quad & \text{Cov}(\hat{\boldsymbol{\theta}}) \ge \mathcal{I}(\boldsymbol{\theta})^{-1} \\
\text{Information Bottleneck:} \quad & \min_{p(t|x)} \left[ I(X; T) - \beta I(T; Y) \right] \\
\text{Expected Information Gain:} \quad & \text{EIG}(d) = I(\boldsymbol{\theta}; Y \mid d) = H(Y \mid d) - \mathbb{E}_{\boldsymbol{\theta}}[H(Y \mid \boldsymbol{\theta}, d)] \\
\text{BALD Active Learning:} \quad & \text{BALD}(x) = H(Y \mid x, \mathcal{D}) - \mathbb{E}_{p(\boldsymbol{\theta} \mid \mathcal{D})}[H(Y \mid x, \boldsymbol{\theta})]
\end{aligned}
$$

### Part B: Bayesian Statistics & Decision-Theoretic Foundations

$$
\begin{aligned}
\text{Posterior Odds & Log-Evidence:} \quad & O_{\text{post}} = O_{\text{prior}} \times \text{LR}, \qquad \Delta \text{Evidence} = \log_2(\text{LR}) \quad \text{[bits]} \\
\text{SLA Tail Risk (Incomplete Beta):} \quad & P(\theta > \tau \mid \mathcal{D}) = 1 - I_\tau(\alpha, \beta) = \frac{1}{B(\alpha, \beta)} \int_\tau^1 \theta^{\alpha-1} (1-\theta)^{\beta-1} d\theta \\
\text{Beta-Binomial Predictive Streak:} \quad & P(\text{pass next } k \mid \alpha, \beta) = \frac{B(\alpha, \beta + k)}{B(\alpha, \beta)} = \prod_{j=0}^{k-1} \frac{\beta + j}{\alpha + \beta + j} \\
\text{Hierarchical Partial Pooling:} \quad & \tilde{\theta}_i = w_i \hat{\theta}_i + (1 - w_i) \mu_{\text{suite}}, \qquad w_i = \frac{n_i}{n_i + M} \\
\text{Bayes Factor & Fix Verification:} \quad & BF_{10} = \frac{p(\mathcal{D} \mid H_1)}{p(\mathcal{D} \mid H_0)}, \qquad P(\theta_B < \theta_A \mid \mathcal{D}) = \mathbb{E}[\mathbb{I}(\theta_B < \theta_A)] \ge 0.95 \\
\text{Asymmetric Bayes Decision Threshold:} \quad & p^* = \frac{C_{\text{false\_block}}}{C_{\text{break}} + C_{\text{false\_block}}} \\
\text{Exponential Forgetting Filter (}\gamma\text{):} \quad & \alpha_t = \gamma \alpha_{t-1} + k_t, \quad \beta_t = \gamma \beta_{t-1} + (m_t - k_t), \quad N_{\text{eff}} = \frac{m}{1 - \gamma} \\
\text{Wald SPRT Stopping Boundaries:} \quad & A = \ln\left(\frac{1-\beta_{\text{err}}}{\alpha_{\text{err}}}\right), \qquad B = \ln\left(\frac{\beta_{\text{err}}}{1-\alpha_{\text{err}}}\right) \\
\text{WAIC Deviance:} \quad & \text{WAIC} = -2 \left( \text{lppd} - \sum_{i=1}^n \text{Var}_{\text{post}}(\ln p(y_i \mid \boldsymbol{\theta})) \right)
\end{aligned}
$$


## 4. The On-Call Operational Action Runbook

When an alert fires in your monitoring system, look up the diagnostic metric below to determine the immediate operational remediation:

| Observed Signal / Diagnostic Metric | Theoretical Meaning | Immediate Operational Action |
| :--- | :--- | :--- |
| **$I(x) \ge 10$ bits (High Surprise)** | Observed failure had $<0.1\%$ prior probability. | **Halt deployment immediately**. Do not retry; this is an unexpected acute regression. |
| **$H_2(\theta) \ge 0.90$ bits (Max Flakiness)** | Test outcome is close to a fair coin toss ($40–60\%$ failure). | **Quarantine test to shadow runner**. Remove from critical merge path until deflaked. |
| **$D_{\text{KL}}(P_{\text{real}} \parallel Q_{\text{prior}}) \ge 2.5$ bits** | Current fleet prior is heavily misaligned with production reality. | **Re-anchor MaxEnt fleet prior**. Recalibrate hyperparameter baseline $\alpha_0, \beta_0$. |
| **$\text{Cond}(\mathcal{I}) \ge 1,000$ (High Condition #)** | Severe multicollinearity in flakiness regression covariates. | **Prune collinear metrics**. Drop redundant host metrics (e.g. keep CPU load, drop thread count). |
| **Shrinkage Factor $B_j \ge 0.80$** | Test has small sample size ($n_j < 5$); posterior pulls to fleet mean. | **Do NOT block PR on single failure**. Dispatch 20 background warm-up runs. |
| **PSIS-LOO Pareto $k_i \ge 0.70$** | Outlier with infinite importance-sampling variance. | **Isolate as Infrastructure Crash**. Route ticket to DevOps/SRE, not PR author. |
| **$\Delta \text{elpd}_{\text{loo}} \le 2 \times \text{SE}$** | Proposed new ML flakiness predictor fails out-of-sample cross-validation. | **Reject the PR / model deployment**. The added complexity is pure overfitting. |
| **SPRT $S_m \le B$** | Cumulative log-likelihood ratio crossed lower boundary. | **Auto-merge pull request**. Flakiness hypothesis dismissed within specified $\beta_{\text{err}}$. |
| **SPRT $S_m \ge A$** | Cumulative log-likelihood ratio crossed upper boundary. | **Auto-reject pull request**. Real regression confirmed within specified $\alpha_{\text{err}}$. |
| **$\max_c p_t(c) \ge 0.95$ (Horstein)** | Posterior probability mass on culprit commit exceeds 95%. | **Auto-revert culprit commit**. Notify author with test failure stack trace. |
| **$P(\theta > \tau_{\text{SLA}}) \ge 0.90$ & $W_{\text{eff}} \le 0.15$** | Posterior SLA tail risk confirmed with narrow epistemic uncertainty. | **Auto-Quarantine Entry**. Move test off blocking merge path to shadow runner. |
| **$P(\text{pass next } 25 \mid D) \ge 0.90$** | Posterior predictive probability of next 25 runs passing exceeds 90%. | **Quarantine Exit (Streak Gate)**. Graduate test back to blocking merge status. |
| **$P(\theta_{\text{post}} < \theta_{\text{pre}} \mid D) \ge 0.95$** | Bayesian A/B hypothesis test confirms post-fix failure rate is lower. | **Verify Fix PR**. Approve de-flaking pull request for merge. |
| **Discounted $D_{\text{KL}} \ge 5.0$ bits** | Parameter distribution shift detected via exponential forgetting. | **Acute Outage Alarm**. Escalate to on-call team; kernel/infra regression in progress. |


## 5. Python Implementation: The Master Curriculum Verification Suite

Let us implement an automated verification test suite that executes sanity checks across both
Information Theory and Bayesian Statistics core formulas:
1. Surprise & Shannon entropy identity verification.
2. KL divergence Gibbs' inequality ($D_{\text{KL}} \ge 0$).
3. Binary entropy curve & maximum at 0.50.
4. Fisher information of Bernoulli distribution.
5. Asymmetric Bayes merge decision threshold $p^*$.
6. Wald SPRT boundary ordering & invariance.
7. Zero-noise clean channel capacity.
8. Posterior odds & log-evidence consistency ($O_{\text{post}} = O_{\text{prior}} \times \text{LR}$).
9. Beta-Binomial posterior predictive streak formula verification ($B(\alpha, \beta+k)/B(\alpha, \beta)$).
10. Discounted Beta-Binomial filter asymptotic memory ($N_{\text{eff}} = \frac{m}{1-\gamma}$).
11. SLA tail risk calculation via incomplete Beta ($1 - I_\tau(\alpha, \beta)$).
12. Horstein probabilistic bisection median invariant check.


```python
import numpy as np
import scipy.stats as stats
import matplotlib.pyplot as plt

np.random.seed(42)

# Set publication style
plt.rcParams.update({
    'font.size': 12,
    'axes.labelsize': 13,
    'axes.titlesize': 14,
    'xtick.labelsize': 11,
    'ytick.labelsize': 11,
    'legend.fontsize': 11,
    'figure.titlesize': 16,
    'figure.dpi': 120
})
```

```python
# -------------------------------------------------------------
# 1. Master Formula Automated Sanity Test Suite
# -------------------------------------------------------------
from scipy.special import betaln, gamma as gamma_fn
from scipy.stats import beta

tests_passed = 0
total_tests = 12

# Test 1: Self-Information & Entropy
p = np.array([0.7, 0.2, 0.1])
H = -np.sum(p * np.log2(p))
I = -np.log2(p)
assert np.isclose(np.sum(p * I), H), "Failed: Expected E[I(x)] == H(X)"
tests_passed += 1

# Test 2: Gibbs' Inequality (KL >= 0)
q = np.array([0.333, 0.333, 0.334])
kl = np.sum(p * np.log2(p / q))
assert kl >= 0.0, "Failed: Gibbs' inequality violated (D_KL < 0)"
tests_passed += 1

# Test 3: Binary Entropy Maximum at 0.5
th = np.linspace(0.01, 0.99, 100)
h2 = -th * np.log2(th) - (1 - th) * np.log2(1 - th)
assert np.isclose(th[np.argmax(h2)], 0.5, atol=0.02), "Failed: Max binary entropy not at 0.5"
tests_passed += 1

# Test 4: Fisher Information of Bernoulli
theta_val = 0.2
I_fisher = 1.0 / (theta_val * (1.0 - theta_val))
assert np.isclose(I_fisher, 6.25), "Failed: Bernoulli Fisher Information mismatch"
tests_passed += 1

# Test 5: Bayes Merge Threshold Formula
C_out, C_blk = 50000.0, 150.0
p_opt = C_blk / (C_out + C_blk)
assert np.isclose(p_opt, 150.0 / 50150.0), "Failed: Bayes threshold calculation mismatch"
tests_passed += 1

# Test 6: Wald SPRT Boundary Invariance
a_err, b_err = 0.01, 0.001
A_bnd = np.log((1 - b_err) / a_err)
B_bnd = np.log(b_err / (1 - a_err))
assert A_bnd > 0 and B_bnd < 0, "Failed: SPRT boundary signs invalid"
tests_passed += 1

# Test 7: Channel Capacity of Zero-Noise Channel
eps_zero = 1e-12
C_clean = 1.0 - (-eps_zero * np.log2(eps_zero) - (1-eps_zero)*np.log2(1-eps_zero))
assert np.isclose(C_clean, 1.0, atol=1e-4), "Failed: Clean channel capacity must be 1 bit"
tests_passed += 1

# Test 8: Posterior Odds & Log-Evidence (LR in bits)
prior_odds = 0.25  # 1:4 prior
lr = 8.0           # Likelihood ratio = 8
post_odds = prior_odds * lr  # 2.0
log_evidence_bits = np.log2(lr)  # 3.0 bits
assert np.isclose(post_odds, 2.0) and np.isclose(log_evidence_bits, 3.0)
tests_passed += 1

# Test 9: Beta-Binomial Posterior Predictive Streak
alpha_val, beta_val, k_streak = 1.0, 100.0, 25
log_streak = betaln(alpha_val, beta_val + k_streak) - betaln(alpha_val, beta_val)
p_streak = np.exp(log_streak)
assert np.isclose(p_streak, 100.0 / 125.0)
tests_passed += 1

# Test 10: Discounted Beta-Binomial Asymptotic Memory N_eff
gamma_disc, m_batch = 0.95, 10
N_eff_formula = m_batch / (1.0 - gamma_disc)
assert np.isclose(N_eff_formula, 200.0)
tests_passed += 1

# Test 11: SLA Tail Risk via Regularized Incomplete Beta
alpha_t, beta_t, tau = 5.0, 95.0, 0.05
tail_risk = 1.0 - beta.cdf(tau, alpha_t, beta_t)
assert 0.0 <= tail_risk <= 1.0
tests_passed += 1

# Test 12: Horstein Probabilistic Bisection Median Invariant
p_horstein = np.array([0.1, 0.4, 0.4, 0.1])
cdf_horstein = np.cumsum(p_horstein)
median_idx = np.argmin(np.abs(cdf_horstein - 0.50))
assert median_idx == 1, "Median index must be commit 1 (c2)"
tests_passed += 1

print(f"MASTER SUITE VERIFIED: {tests_passed} / {total_tests} Theoretical Identities Validated!")
```

```python
# -------------------------------------------------------------
# 2. Master Curriculum Visual Summary
# -------------------------------------------------------------
fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(16, 12))

# Subplot 1: Binary Entropy Curve (Ch 01)
ax1.plot(th * 100, h2, color='teal', linewidth=2.5)
ax1.axvline(50, color='crimson', linestyle='--', label='Max Flakiness (1.0 bit)')
ax1.set_xlabel('Flakiness Probability (%)')
ax1.set_ylabel('Shannon Entropy (bits)')
ax1.set_title('1. Flakiness Uncertainty Curve (Ch 01)')
ax1.grid(True, alpha=0.3)
ax1.legend()

# Subplot 2: Forward vs Reverse KL Asymmetry (Ch 02)
x_axis = np.linspace(-4, 4, 200)
p_dist = 0.5 * stats.norm.pdf(x_axis, -1.5, 0.6) + 0.5 * stats.norm.pdf(x_axis, 1.5, 0.6)
q_mode = stats.norm.pdf(x_axis, -1.5, 0.8) # Reverse KL mode-seeking
q_mean = stats.norm.pdf(x_axis, 0.0, 1.8)  # Forward KL mean-seeking
ax2.plot(x_axis, p_dist, 'k-', linewidth=2, label='True Multimodal Flakiness P')
ax2.plot(x_axis, q_mean, 'r--', linewidth=2, label='Forward KL: Mean-Seeking')
ax2.plot(x_axis, q_mode, 'b:', linewidth=2, label='Reverse KL: Mode-Seeking')
ax2.set_title('2. Relative Entropy Asymmetry (Ch 02)')
ax2.grid(True, alpha=0.3)
ax2.legend()

# Subplot 3: Fisher Information Curvature (Ch 05)
ax3.plot(th * 100, 1.0 / (th * (1.0 - th)), color='darkorange', linewidth=2.5)
ax3.set_xlabel('Parameter $\\theta$ (%)')
ax3.set_ylabel('Fisher Information $\\mathcal{I}(\\theta)$')
ax3.set_title('3. Information Geometric Curvature (Ch 05)')
ax3.grid(True, alpha=0.3)

# Subplot 4: SPRT Decision Boundaries (Ch 09)
steps_axis = np.arange(15)
ax4.axhline(A_bnd, color='crimson', linestyle='--', linewidth=2, label='Boundary A (Reject PR)')
ax4.axhline(B_bnd, color='teal', linestyle='--', linewidth=2, label='Boundary B (Safe Merge)')
# Sample random walk
rw = np.cumsum(np.random.choice([0.65, -0.45], size=15, p=[0.7, 0.3]))
ax4.plot(steps_axis, rw, 'o-', color='purple', linewidth=2, label='Evidence Walk $S_m$')
ax4.set_xlabel('Sequential Retries $m$')
ax4.set_ylabel('Log-Likelihood Ratio $S_m$ (nats)')
ax4.set_title('4. Wald SPRT Sequential Stopping (Ch 09)')
ax4.grid(True, alpha=0.3)
ax4.legend()

plt.tight_layout()
plt.show()
```

```python
# -------------------------------------------------------------
# Exercise Tier 1 Solution: Master Unit Converter
# -------------------------------------------------------------
# 1 bit = ln(2) nats approx 0.693147 nats
# 1 nat = log2(e) bits = 1 / ln(2) approx 1.442695 bits

bits_val = 2.5
nats_val = bits_val * np.log(2)

nats_input = 4.0
bits_output = nats_input / np.log(2)

L_bits = 6
implied_q = 2.0**(-L_bits)

print("--- Tier 1: Master Unit Converter Results ---")
print(f"2.5 bits  = {nats_val:.4f} nats")
print(f"4.0 nats  = {bits_output:.4f} bits")
print(f"L=6 bits  = implied probability q = {implied_q:.6f} ({implied_q*100:.2f}%)")
```

```python
# -------------------------------------------------------------
# Exercise Tier 2 Solution: Multi-Service Incident Triage Bot
# -------------------------------------------------------------
incidents = [
    {"service": "Service A", "surprise_bits": 11.2, "pareto_k": 0.35},
    {"service": "Service B", "surprise_bits": 2.1,  "pareto_k": 0.85},
    {"service": "Service C", "surprise_bits": 8.4,  "pareto_k": 0.22}
]

print("\n--- Tier 2: Multi-Service Incident Triage Bot ---")
for inc in incidents:
    s_name = inc["service"]
    bits = inc["surprise_bits"]
    k = inc["pareto_k"]
    
    if k >= 0.70:
        decision = "INFRASTRUCTURE_CRASH -> Route to DevOps/SRE (Worker Outlier)"
    elif bits >= 8.0:
        decision = "BLOCKING_REGRESSION  -> Block Merge & Alert Developer (Acute Bug)"
    else:
        decision = "ROUTINE_FLAKE        -> Safe to Retry / Shadow Quarantine"
        
    print(f"{s_name:<10}: Surprise = {bits:>4.1f} bits | Pareto k = {k:.2f} | Triage: {decision}")
```

```python
# -------------------------------------------------------------
# Exercise Tier 3 Solution: Interactive 11-Point Telemetry Card
# -------------------------------------------------------------
print("\n--- Tier 3: Interactive Telemetry Card Generator ---")

def generate_telemetry_card(test_name, k, n, latest_y, C_outage=50000.0, C_block=150.0):
    mle = k / n
    alpha_post = 1.2 + k
    beta_post = 40.0 + n - k
    post_mean = alpha_post / (alpha_post + beta_post)
    h2 = -post_mean * np.log2(post_mean) - (1 - post_mean) * np.log2(1 - post_mean)
    surprise = -np.log2(post_mean if latest_y == 1 else (1.0 - post_mean))
    p_star = C_block / (C_outage + C_block)
    
    lines = [
        "=" * 70,
        f"TEST RELIABILITY & INFORMATION SCORECARD: {test_name}",
        "=" * 70,
        f"Historical Runs (n):         {n:,} runs ({k} failures, MLE = {mle*100:.2f}%)",
        f"Shrunk Posterior Rate:      {post_mean*100:.3f}% (Empirical Bayes Partial Pooling)",
        f"Flakiness Entropy H2:       {h2:.4f} bits / run (Channel Capacity C = {1.0-h2:.4f} bits)",
        f"Latest Outcome Surprise:    {surprise:.2f} bits of information emitted",
        f"Bayes Merge Gate (p*):      {p_star*100:.3f}% (Based on ${C_outage:,.0f} outage SLA)",
        f"Recommended Action:         {'QUARANTINE / BLOCK' if post_mean > p_star else 'SAFE TO MERGE'}",
        "=" * 70
    ]
    return "\n".join(lines)

print(generate_telemetry_card("PaymentCheckoutIntegrationTest", 3, 400, 1))
```
