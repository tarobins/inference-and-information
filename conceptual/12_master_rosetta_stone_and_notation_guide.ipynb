## 2. Notation Collision Survival Guide

When reading interdisciplinary literature, statistical libraries (PyMC, ArviZ,
SciPy), or textbooks (Cover & Thomas vs McElreath vs MacKay), several
mathematical symbols collide disastrously. Use this guide to de-conflict them:

### Collision 1: The Letter $H$
- $H(X)$: **Shannon Entropy** (Information Theory) -- scalar in bits or nats.
- $\mathbf{H}$ or $H_{ij}$: **Hessian Matrix** (Calculus / Bayesian Laplace) --
  matrix of second-order partial derivatives $\frac{\partial^2 \ln p}{\partial \theta_i \partial \theta_j}$.
- $H_0, H_1$: **Null and Alternative Hypotheses** (Decision Theory / Wald SPRT).
- $h(X)$: **Differential Entropy** of a continuous random variable.

### Collision 2: The Letter $I$
- $I(x)$: **Self-Information / Surprise** of an outcome $x$ (scalar bits).
- $I(X; Y)$: **Mutual Information** between random variables $X$ and $Y$ (scalar bits).
- $\mathcal{I}(\boldsymbol{\theta})$: **Fisher Information Matrix** (Information
  Geometry) -- positive semi-definite Riemannian metric tensor.
- $\mathbf{I}$: **Identity Matrix** ($\mathbf{I}_n$).
- $I_x(a, b)$: **Regularized Incomplete Beta Function** (Cumulative Beta CDF).

### Collision 3: The Letter $D$
- $D_{\text{KL}}(P \parallel Q)$: **Kullback-Leibler Divergence** (Relative Entropy).
- $\mathcal{D}$: **Observed Dataset** ($\{\mathbf{y}_1, \dots, \mathbf{y}_n\}$).
- $R(D)$: **Rate-Distortion Function** ($D$ is the maximum distortion budget).
- $d$: **Experimental Design Action** (Lindley EIG).

### Collision 4: The Greek Letter $\beta$
- $\text{Beta}(\alpha, \beta)$: Second shape parameter of the Beta distribution
  (prior pseudo-successes / failures).
- $\beta$ in Information Bottleneck: The Lagrange multiplier $\min I(X; T) - \beta I(T; Y)$
  controlling the compression-retention exchange rate.
- $\beta$ in MCMC / Physics: Inverse temperature $\beta = \frac{1}{k_B T}$ used
  in simulated annealing and parallel tempering.
- $\beta_{\text{err}}$: Type II error probability (probability of merging a real
  regression) in Wald SPRT.

### Collision 5: The Greek Letter $\theta$
- $\theta$: Bernoulli success / failure probability ($\theta \in [0, 1]$).
- $\boldsymbol{\theta}$: General parameter vector ($\boldsymbol{\theta} \in \mathbb{R}^D$).
- $\Theta_0, \Theta_1$: Discrete hypothesis parameter regions in decision theory.


## 3. Master Formula Cheat Sheet

### Part A: Information-Theoretic Foundations

$$
\begin{aligned}
\text{Self-Information (Surprise):} \quad & I(x) = -\log_2 P(x) \quad \text{[bits]} \\
\text{Shannon Entropy:} \quad & H(X) = -\sum_{x} p(x) \log_2 p(x) \\
\text{Binary Entropy:} \quad & H_2(\theta) = -\theta \log_2(\theta) - (1-\theta) \log_2(1-\theta) \\
\text{Relative Entropy (KL):} \quad & D_{\text{KL}}(P \parallel Q) = \sum_{x} P(x) \log_2 \frac{P(x)}{Q(x)} \ge 0 \\
\text{Cross-Entropy:} \quad & H(P, Q) = H(P) + D_{\text{KL}}(P \parallel Q) \\
\text{Mutual Information:} \quad & I(X; Y) = H(X) - H(X \mid Y) = H(Y) - H(Y \mid X) \\
\text{Fisher Information Matrix:} \quad & \mathcal{I}_{ij}(\boldsymbol{\theta}) = \mathbb{E}\left[ \left(\frac{\partial \ln p}{\partial \theta_i}\right) \left(\frac{\partial \ln p}{\partial \theta_j}\right) \right] = - \mathbb{E}\left[ \frac{\partial^2 \ln p}{\partial \theta_i \partial \theta_j} \right] \\
\text{Cramér-Rao Lower Bound:} \quad & \text{Cov}(\hat{\boldsymbol{\theta}}) \ge \mathcal{I}(\boldsymbol{\theta})^{-1} \\
\text{Information Bottleneck:} \quad & \min_{p(t|x)} \left[ I(X; T) - \beta I(T; Y) \right] \\
\text{Expected Information Gain:} \quad & \text{EIG}(d) = I(\boldsymbol{\theta}; Y \mid d) = H(Y \mid d) - \mathbb{E}_{\boldsymbol{\theta}}[H(Y \mid \boldsymbol{\theta}, d)] \\
\text{BALD Active Learning:} \quad & \text{BALD}(x) = H(Y \mid x, \mathcal{D}) - \mathbb{E}_{p(\boldsymbol{\theta} \mid \mathcal{D})}[H(Y \mid x, \boldsymbol{\theta})]
\end{aligned}
$$

### Part B: Bayesian Statistics & Decision-Theoretic Foundations

$$
\begin{aligned}
\text{Posterior Odds & Log-Evidence:} \quad & O_{\text{post}} = O_{\text{prior}} \times \text{LR}, \qquad \Delta \text{Evidence} = \log_2(\text{LR}) \quad \text{[bits]} \\
\text{SLA Tail Risk (Incomplete Beta):} \quad & P(\theta > \tau \mid \mathcal{D}) = 1 - I_\tau(\alpha, \beta) = \frac{1}{B(\alpha, \beta)} \int_\tau^1 \theta^{\alpha-1} (1-\theta)^{\beta-1} d\theta \\
\text{Beta-Binomial Predictive Streak:} \quad & P(\text{pass next } k \mid \alpha, \beta) = \frac{B(\alpha, \beta + k)}{B(\alpha, \beta)} = \prod_{j=0}^{k-1} \frac{\beta + j}{\alpha + \beta + j} \\
\text{Hierarchical Partial Pooling:} \quad & \tilde{\theta}_i = w_i \hat{\theta}_i + (1 - w_i) \mu_{\text{suite}}, \qquad w_i = \frac{n_i}{n_i + M} \\
\text{Bayes Factor & Fix Verification:} \quad & BF_{10} = \frac{p(\mathcal{D} \mid H_1)}{p(\mathcal{D} \mid H_0)}, \qquad P(\theta_B < \theta_A \mid \mathcal{D}) = \mathbb{E}[\mathbb{I}(\theta_B < \theta_A)] \ge 0.95 \\
\text{Asymmetric Bayes Decision Threshold:} \quad & p^* = \frac{C_{\text{false\_block}}}{C_{\text{break}} + C_{\text{false\_block}}} \\
\text{Exponential Forgetting Filter (}\gamma\text{):} \quad & \alpha_t = \gamma \alpha_{t-1} + k_t, \quad \beta_t = \gamma \beta_{t-1} + (m_t - k_t), \quad N_{\text{eff}} = \frac{m}{1 - \gamma} \\
\text{Wald SPRT Stopping Boundaries:} \quad & A = \ln\left(\frac{1-\beta_{\text{err}}}{\alpha_{\text{err}}}\right), \qquad B = \ln\left(\frac{\beta_{\text{err}}}{1-\alpha_{\text{err}}}\right) \\
\text{WAIC Deviance:} \quad & \text{WAIC} = -2 \left( \text{lppd} - \sum_{i=1}^n \text{Var}_{\text{post}}(\ln p(y_i \mid \boldsymbol{\theta})) \right)
\end{aligned}
$$


## 4. The On-Call Operational Action Runbook

When an alert fires in your monitoring system, look up the diagnostic metric below to determine the immediate operational remediation:

| Observed Signal / Diagnostic Metric | Theoretical Meaning | Immediate Operational Action |
| :--- | :--- | :--- |
| **$I(x) \ge 10$ bits (High Surprise)** | Observed failure had $<0.1\%$ prior probability. | **Halt deployment immediately**. Do not retry; this is an unexpected acute regression. |
| **$H_2(\theta) \ge 0.90$ bits (Max Flakiness)** | Test outcome is close to a fair coin toss ($40–60\%$ failure). | **Quarantine test to shadow runner**. Remove from critical merge path until deflaked. |
| **$D_{\text{KL}}(P_{\text{real}} \parallel Q_{\text{prior}}) \ge 2.5$ bits** | Current fleet prior is heavily misaligned with production reality. | **Re-anchor MaxEnt fleet prior**. Recalibrate hyperparameter baseline $\alpha_0, \beta_0$. |
| **$\text{Cond}(\mathcal{I}) \ge 1,000$ (High Condition #)** | Severe multicollinearity in flakiness regression covariates. | **Prune collinear metrics**. Drop redundant host metrics (e.g. keep CPU load, drop thread count). |
| **Shrinkage Factor $B_j \ge 0.80$** | Test has small sample size ($n_j < 5$); posterior pulls to fleet mean. | **Do NOT block PR on single failure**. Dispatch 20 background warm-up runs. |
| **PSIS-LOO Pareto $k_i \ge 0.70$** | Outlier with infinite importance-sampling variance. | **Isolate as Infrastructure Crash**. Route ticket to DevOps/SRE, not PR author. |
| **$\Delta \text{elpd}_{\text{loo}} \le 2 \times \text{SE}$** | Proposed new ML flakiness predictor fails out-of-sample cross-validation. | **Reject the PR / model deployment**. The added complexity is pure overfitting. |
| **SPRT $S_m \le B$** | Cumulative log-likelihood ratio crossed lower boundary. | **Auto-merge pull request**. Flakiness hypothesis dismissed within specified $\beta_{\text{err}}$. |
| **SPRT $S_m \ge A$** | Cumulative log-likelihood ratio crossed upper boundary. | **Auto-reject pull request**. Real regression confirmed within specified $\alpha_{\text{err}}$. |
| **$\max_c p_t(c) \ge 0.95$ (Horstein)** | Posterior probability mass on culprit commit exceeds 95%. | **Auto-revert culprit commit**. Notify author with test failure stack trace. |
| **$P(\theta > \tau_{\text{SLA}}) \ge 0.90$ & $W_{\text{eff}} \le 0.15$** | Posterior SLA tail risk confirmed with narrow epistemic uncertainty. | **Auto-Quarantine Entry**. Move test off blocking merge path to shadow runner. |
| **$P(\text{pass next } 25 \mid D) \ge 0.90$** | Posterior predictive probability of next 25 runs passing exceeds 90%. | **Quarantine Exit (Streak Gate)**. Graduate test back to blocking merge status. |
| **$P(\theta_{\text{post}} < \theta_{\text{pre}} \mid D) \ge 0.95$** | Bayesian A/B hypothesis test confirms post-fix failure rate is lower. | **Verify Fix PR**. Approve de-flaking pull request for merge. |
| **Discounted $D_{\text{KL}} \ge 5.0$ bits** | Parameter distribution shift detected via exponential forgetting. | **Acute Outage Alarm**. Escalate to on-call team; kernel/infra regression in progress. |
